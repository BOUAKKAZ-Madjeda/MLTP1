```python
DATAPATH = "./txt_file.txt"
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_20newsgroups
from collections import Counter
import re
```

```python
newsgroups = fetch_20newsgroups(subset='train', shuffle=True, random_state=42)
```

```python
words = re.findall("[a-z]+", " ".join(newsgroups.data).lower())
print(f"Number of words: {len(words)}")
```

```python
def find_word(word, documents=newsgroups.data, labels=newsgroups.target,
              class_names=newsgroups.target_names, max_results=None):
    pattern = re.compile(rf"\b{re.escape(word)}\b", re.IGNORECASE)
    results = []
    for i, doc in enumerate(documents):
        count = len(pattern.findall(doc))
        if count > 0:
            results.append((i, class_names[labels[i]], count))
            if max_results and len(results) >= max_results:
                break
    return results

print(find_word("MONEy"))
```
