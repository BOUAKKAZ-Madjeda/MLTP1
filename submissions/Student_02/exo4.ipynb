```python
DATA = "./matrix.csv"
import pandas as pd
import numpy as np
```

```python
matrix = pd.read_csv(DATA)
matrix
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, MaxAbsScaler, RobustScaler, Binarizer

print("Standard Scaling:")
print(StandardScaler().fit_transform(matrix))
print("\n")

print("Min Max Scaling:")
print(MinMaxScaler().fit_transform(matrix))
print("\n")

print("Max Abs Scaling:")
print(MaxAbsScaler().fit_transform(matrix))
print("\n")

print("Robust Scaling:")
print(RobustScaler().fit_transform(matrix))
print("\n")

print("Normalizing:")
print(matrix / np.linalg.norm(matrix))
```


| Method	| Formula	| Works on | Resulting range / property |
| --- | --- | --- | --- |
| StandardScaler	| (x − mean) / std	| each feature (column)	| mean 0, std 1 per column; unbounded |
| MinMaxScaler	| (x − min) / (max − min)	| each feature	| [0, 1] per column |
| MaxAbsScaler	| x / max(‖x‖)	| each feature	| [−1, 1]; keeps zeros and sparsity, no shifting |
| RobustScaler	| (x − median) / IQR	| each feature	| median 0; uses quartiles, so outliers have less influence |
| normalize	| x / ‖x‖ | (L2 by default)	| each sample (row)	| every row has length 1 |

```python
thresholds = [0.1, 0.2, 0.3, 0.4, 0.6, 0.8]

for threshold in thresholds:
    print(f"Threshold = {threshold}")
    print(Binarizer(threshold=threshold).fit_transform(matrix))
    print("\n")
```

```python
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder, OneHotEncoder

X = np.array([["male",   "from US",     "uses Safari"],
              ["female", "from Europe", "uses Safari"],
              ["female", "from Asia",   "uses Firefox"],
              ["male",   "from Africa", "uses Chrome"]])
y = np.array(["US", "France", "Japan", "Algeria"])
```

```python
enc_X = OrdinalEncoder()
X_enc = enc_X.fit_transform(X)
print(enc_X.categories_)
print(X_enc)
```

```python
le = LabelEncoder()
y_enc = le.fit_transform(y)
print(y_enc)
for i, c in enumerate(le.classes_):
    print(c, "->", i)
```

```python
ohe = OneHotEncoder(sparse_output=False)
y_onehot = ohe.fit_transform(y.reshape(-1, 1))
print(ohe.categories_)
print(y_onehot)
```
